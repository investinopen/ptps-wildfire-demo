# Proxy demo

This shows examples of leveraging the proxy from different tools. These cases are mirrored in the tests.

First, [start the proxy](../README.md#option-1-http-proxy).


## Configuration

See [`constants.py`](constants.py).


```python
from constants import CERT_PATH, PROXY_URL
```

## [Requests](https://docs.python-requests.org/)

Without the proxy:


```python
import requests

response = requests.get("https://www.epa.gov/ejscreen", timeout=10)

print(response)
response.text
```

With the proxy:


```python
proxies = {"http": PROXY_URL, "https": PROXY_URL}
response = requests.get(
    "https://www.epa.gov/ejscreen", proxies=proxies, verify=str(CERT_PATH), timeout=10
)

print(response)
response.text
```
