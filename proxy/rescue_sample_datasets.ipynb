```python
import asyncio

import httpx

from ptps_wildfire_demo.proxy.resolver import Resolver

# All datasets that don't require a key. URLs derived via Copilot.
urls = [
    "https://data.source.coop/cboettig/fire/usgs-mtbs.parquet",
    "https://data-usfs.hub.arcgis.com/datasets/usfs::probabilistic-wildfire-risk-burn-probability-image-service/explore?location=45.200890%2C-122.025647%2C2",
    "https://wildfirerisk.org/download/",
    "https://source.coop/carbonplan/carbonplan-ocr",
    "https://source.coop/cboettig/fire",
    "https://source.coop/bkr/obs",
    "https://source.coop/giswqs/nwi",
    "https://api.weather.gov/alerts/active?event=Red%20Flag%20Warning",
    "https://www.ospo.noaa.gov/products/land/hms.html",
    "https://www.ncdc.noaa.gov/stormevents/",
    "https://firms.modaps.eosdis.nasa.gov/",
    "https://data-nifc.opendata.arcgis.com/",
    "https://gis.blm.gov/arcgis/rest/services/fire/BLM_Natl_FirePerimeter/MapServer",
    "https://modis.gsfc.nasa.gov/data/dataprod/mod13.php",
    "https://dataspace.copernicus.eu/",
    "https://overpass-api.de/",
    "https://www.landfire.gov/",
    "https://www.landfire.gov/fuel/seasonal_fuels",
    "https://www.cdc.gov/places/",
    "https://www.atsdr.cdc.gov/place-health/php/svi/index.html",
    "https://www.epa.gov/ejscreen",
    "https://www.osti.gov/dataexplorer/biblio/dataset/1975203",
    "https://doi.org/10.5061/dryad.63xsj3vd4",
    "https://doi.org/10.5061/dryad.0zpc867bw",
]

client = httpx.AsyncClient()
resolver = Resolver(client)
```

```python
rescues = await asyncio.gather(*(resolver.get_rescue(url) for url in urls))
# rescues
```

```python
import pandas as pd

pd.set_option("display.max_colwidth", 200)
pd.DataFrame(rescues)
```
