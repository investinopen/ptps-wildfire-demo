```python
import pandas as pd
from helpers import render_links

terminations = pd.read_csv(
    "https://dataindex.us/terminations-tracker/download.csv",
    dtype_backend="pyarrow",
)
render_links(terminations.head())
```

```python
terminations["Section"].unique()
```

```python
wayback_archived = terminations[terminations["Primary Link Wayback"].notna()]
primary_urls = wayback_archived["Primary Link Wayback"].str.replace(
    r"https?://web\.archive\.org/web/\d+/", "", regex=True
)

terminations["Primary Link"] = primary_urls

render_links(primary_urls.to_frame())
```

```python
import httpx

from ptps_wildfire_demo import Resolver

async with httpx.AsyncClient() as client:
    resolver = Resolver(client)
```

```python
drp_rescues = resolver.drp_rescues

# pull out the rescue URLs
drp_rescues = drp_rescues.assign(
    resource_urls=drp_rescues["resources"].apply(
        lambda resources: ", ".join(
            resource["url"] for resource in resources if resource["url"]
        )
    )
)
drp_rescues = drp_rescues.drop(columns=["resources"])

render_links(drp_rescues.head())
```

## Fuzzy matches


```python
import recordlinkage

MATCH_THRESHOLD = 0.86

indexer = recordlinkage.Index()
indexer.sortedneighbourhood(
    left_on="Dataset Name",
    right_on="title",
    window=5,
)
candidate_pairs = indexer.index(terminations, drp_rescues)

compare = recordlinkage.Compare()
compare.string("Dataset Name", "title", method="jarowinkler", label="title_score")
compare.string("Agency", "agency", method="jarowinkler", label="agency_score")
compare.string(
    "Agency", "organization", method="jarowinkler", label="organization_score"
)

scores = compare.compute(candidate_pairs, terminations, drp_rescues)
scores["agency_score"] = scores[["agency_score", "organization_score"]].max(axis=1)
scores["match_score"] = 0.7 * scores["title_score"] + 0.3 * scores["agency_score"]

best_matches = (
    scores.sort_values("match_score", ascending=False)
    .groupby(level=0)
    .head(1)
    .query("match_score >= @MATCH_THRESHOLD")
    .reset_index(names=["termination_index", "rescue_index"])
)

title_matches = best_matches.merge(
    terminations[
        [
            "Dataset Name",
            "Agency",
            "Link 1 URL",
            "Link 2 URL",
            "Link 3 URL",
            "Primary Link",
        ]
    ],
    left_on="termination_index",
    right_index=True,
).merge(
    drp_rescues[["title", "agency", "organization", "data_source", "resource_urls"]],
    left_on="rescue_index",
    right_index=True,
)

render_links(
    title_matches[
        [
            "Dataset Name",
            "Agency",
            "title",
            "agency",
            "organization",
            "title_score",
            "agency_score",
            "match_score",
            "Link 1 URL",
            "Link 2 URL",
            "Link 3 URL",
            "Primary Link",
            "data_source",
            "resource_urls",
        ]
    ].sort_values("match_score", ascending=False)
)
```
